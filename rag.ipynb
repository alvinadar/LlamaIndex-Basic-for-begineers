```python
import asyncio
import time
```

```python
# 1. Define asynchronous tasks using 'async def'

async def make_coffee():
    print("Starting to brew coffee...")
    await asyncio.sleep(3)  # Simulate time taken to brew coffee
    print("Coffee is ready!")

async def make_toast():
    print("Starting to toast bread...")
    await asyncio.sleep(2)  # Simulate time taken to toast bread    
    print("Toast is ready!")

async def main():
    start_time = time.time()
    print("Starting breakfast preparation...")

    #Run both tasks concurrently using asyncio.gather
    await asyncio.gather(
        make_coffee(),
        make_toast()
    )

    elapsed_time = time.time() - start_time
    print(f"Breakfast is ready! Total time taken: {elapsed_time:.2f}")

await main()  # Run the main function to start the asynchronous tasks
```
